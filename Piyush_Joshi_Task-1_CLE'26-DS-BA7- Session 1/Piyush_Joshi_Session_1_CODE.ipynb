***Outlier Detection and Feature Engineering***

````python
# Creating Vote Share (Percentage) Features :
years = ['2008','2012','2016']
for y in years:
    df[f'dem_pct_{y}'] = df[f'dem_{y}'] / df[f'total_{y}']
    df[f'gop_pct_{y}'] = df[f'gop_{y}'] / df[f'total_{y}']
    df[f'oth_pct_{y}'] = df[f'oth_{y}'] / df[f'total_{y}']
````

````python
# Vote Margin Feature : Who won and by how much ?
for y in years:
    df[f'margin_{y}'] = df[f'dem_pct_{y}'] - df[f'gop_pct_{y}']
````

````python
# Creating Relative Party Strength :
for y in years:
    df[f'dem_gop_ratio_{y}'] = df[f'dem_{y}'] / (df[f'gop_{y}'] + 1)
````

````python
new_cols = ['dem_pct_2008','gop_pct_2008','oth_pct_2008',
            'dem_pct_2012','gop_pct_2012','oth_pct_2012',
            'dem_pct_2016','gop_pct_2016','oth_pct_2016',
            'margin_2008','margin_2012','margin_2016',
            'dem_gop_ratio_2008','dem_gop_ratio_2012','dem_gop_ratio_2016']
````
